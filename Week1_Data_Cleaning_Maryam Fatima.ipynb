### Day 2: **Identifying and Handling Missing Values**

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\New Computer City\Week 1 Data Cleaning Project\applicants.csv")
cleaned_df = df.copy()
rows_before = df.shape[0]
print(df)
```

This dataset contains missing values that may affect data quality and the reliability of future analysis. The objective is to identify missing values across columns, understand their extent and distribution, and decide how to handle them appropriately using Python and Pandas.

```python
# Remove rows with missing values in the "Email" column
cleaned_df = cleaned_df.dropna(subset=["Email"])
print(cleaned_df)
```

while exploring the dataset I found 6 rows with missing emails. So I dropped 6 rows as the email is the main contact method.

```python
# remove rows with missing values in the "Applicant Name" column
cleaned_df = cleaned_df.dropna(subset=["Applicant Name"])
print(cleaned_df)
```

I found 3 rows with missing values in the "Applicant Name" column so i dropped 3 rows as the name as well as the email are the main contact method.

```python
# filling missing values in the "University" column with "Not Specified"
cleaned_df["University"] = cleaned_df["University"].fillna("Not Specified")
print(cleaned_df)
```

FIlled out the row with missing "University" column with "Not Specified" as university is not the necessery/must element so i kept them to avoid losing useful information.

```python
# filling out the missing values in the "Status" column with "Under Review"
cleaned_df["Status"] = cleaned_df["Status"].fillna("Under Review")
print(cleaned_df)
```

Filled with "Under Review". A blank status most likely means
the application hasn't been processed yet.

```python
print("Rows Removed:", rows_before - cleaned_df.shape[0])
cleaned_df.isnull().sum()
```

**Result:** The dataset went from 86 rows to 77 (9 rows removed). No missing values remain.

### Day 3: **Remove Duplicate Applicants:**
There are 2 kinds of duplicates:
1- Exact Duplicates: Every column is identical so drop_duplicates() catches them.
2- Near Duplicates: The same person with slight different formatting means ali@mail.com and ALI@GMAIL.COM. plain drop_duplicates() catches them.

```python
# Check exact duplicates first
print(cleaned_df.duplicated().sum())
# Remove exact duplicate rows
cleaned_df = cleaned_df.drop_duplicates()
print(cleaned_df)
```

Found and Removed 3 exact duplicate rows containing same information.

```python
print(cleaned_df["Email"].duplicated().sum())
```

```python
cleaned_df["Email"] = cleaned_df["Email"].str.strip().str.lower()
print(cleaned_df)
cleaned_df.duplicated().sum()
```

```python
cleaned_df[cleaned_df['Email'].duplicated(keep=False)].sort_values('Email')
```

```python
cleaned_df = cleaned_df.drop_duplicates(subset=["Email"], keep="first")
print(cleaned_df)
```

Found and Dropped 3 near duplicate rows.

```python
print(cleaned_df.duplicated().sum())
print(cleaned_df['Email'].duplicated().sum())
```

**Result:** The dataset went from 77 rows to 71 rows. No duplicated values remain.

### Day 4: **Standardize Text Fields**

The dataset contains inconsistent capitalization, spelling variations, and spacing in text fields. The goal is to standardize these values to improve data consistency and reliability.

**Steps:**
1. Inspect unique values in categorical columns.
2. Remove leading and trailing spaces.
3. Standardize capitalization.
4. Correct inconsistent spelling and equivalent category names.
5. Verify the cleaned values and compare them with the original data.

**Objective:** Ensure consistent text values for accurate analysis and reporting.

**Standardize Domain Applied:**

```python
cleaned_df["Domain Applied"].unique()
```

```python
cleaned_df["Domain Applied"] = cleaned_df["Domain Applied"].str.strip().str.lower()
# print(cleaned_df)
cleaned_df["Domain Applied"].unique()
```

```python
domain_mapping = {"data science": "Data Science",
                  "datascience": "Data Science",
                  "graphic designing": "Graphic Design",
                  "graphic design": "Graphic Design",
                  "web development": "Web Development",
                  "webdev": "Web Development",
                  "web dev": "Web Development",
                  "mobile app development": "Mobile App Development",
                  "mobile development": "Mobile App Development",
                  "mobile app dev": "Mobile App Development",
                  "digital marketing": "Digital Marketing",
                  "digital mktg": "Digital Marketing"
                }

cleaned_df["Domain Applied"] = cleaned_df["Domain Applied"].map(domain_mapping)
print(cleaned_df["Domain Applied"].isnull().sum())   # should be 0
cleaned_df["Domain Applied"].unique()
```

```python
print(cleaned_df)
```

**Domain Applied:** The column had 12+ spelling and casing variations for only
5 real domains (e.g. "web dev", "WEB DEV", "webdev"). I stripped whitespace and
lowercased the values, then mapped every variation to one of 5 standard labels
using a dictionary. After mapping, `.isnull().sum()` is 0 and `.unique()` shows
exactly 5 values, so no variation was missed.

**Standardize Status:**

```python
cleaned_df["Status"] = cleaned_df["Status"].str.strip().str.title()
cleaned_df["Status"].unique()
```

```python
print(cleaned_df)
```

**Status:** standardized with `.str.strip().str.title()` into 3 values
(Under Review, Selected, Rejected).

**Standardize Applicant Name:**

```python
cleaned_df["Applicant Name"] = cleaned_df["Applicant Name"].str.strip().str.title()
print(cleaned_df)
```

**Applicant Name:** stripped extra spaces and converted to Title Case.

**Standardize University Names:**

```python
cleaned_df["University"] = cleaned_df["University"].str.strip().str.lower()
cleaned_df["University"].unique()
```

```python
university_mapping = {"fast nuces": "FAST NUCES",
                      "fast-nuces": "FAST NUCES",
                      "iba karachi": "IBA Karachi",
                      "ned": "NED University",
                      "ned university": "NED University",
                      "nust": "NUST",
                      "n.u.s.t": "NUST",
                      "uok": "University of Karachi",
                      "university of karachi": "University of Karachi",
                      "comsats": "COMSATS University",
                      "comsats university": "COMSATS University",
                      "lums": "LUMS",
                      "not specified": "Not Specified"
                      }

cleaned_df["University"] = cleaned_df["University"].map(university_mapping)
print(cleaned_df["University"].isnull().sum())   # should be 0            
```

```python
print(cleaned_df)
```

**University:** mapped university to get the exact university names.

**Result:**
Domain Applied went from 18 spelling/casing variations to exactly
5 standard domains (0 missing after mapping). Status now has 3 consistent
values (Under Review, Selected, Rejected), and Applicant Name is in consistent
Title Case. No rows were removed in this step: the dataset is still 71 rows x 7 columns.

### Day 5: **Fix Data Types (Especially Dates and Phone Numbers)**

**The problem:** A date or phone number that's stored as plain text looks right to a human but can't
be sorted, filtered, or calculated on correctly by code until it's converted to the proper type.

**Step 1: Phone numbers**

```python
cleaned_df["Phone"] = cleaned_df["Phone"].str.replace(" ", "", regex=False).str.replace("-", "", regex=False)
print(cleaned_df)
```

```python
print(cleaned_df["Phone"].str.len().value_counts())
```

**Phone:**
Kept as text (phone numbers are identifiers, and leading zeros must
be preserved). Removed dashes and spaces with `.str.replace()`. All 71 numbers
are now 11 digits starting with 03. I did not add country codes because the
data is all local Pakistani numbers and the task needs consistent formatting only.

**Step 2: Dates**

```python
# 1. Back up the original text, BEFORE converting
raw_dates = cleaned_df["Application Date"].copy()
```

```python
# 2. Missing dates before conversion
print(cleaned_df["Application Date"].isnull().sum())
```

```python
# 3. Convert text to real dates (unreadable values become NaT)
cleaned_df["Application Date"] = pd.to_datetime(cleaned_df["Application Date"], errors = "coerce", format = "mixed", dayfirst = True)
print(cleaned_df)
```

```python
# 4. Missing dates after conversion
print(cleaned_df["Application Date"].isnull().sum())
```

```python
# 5. See which rows they are and what was originally written
cleaned_df[cleaned_df["Application Date"].isnull()].assign(original=raw_dates)
```

```python
# 6. Confirm the type and the row count
cleaned_df.info()                                          
print(cleaned_df.shape)
```

**Application Date**

The column held three text formats (2026-08-22, 27/05/2026, 20-08-2026). I
converted it with `pd.to_datetime(errors="coerce", format="mixed",
dayfirst=True)` and confirmed with `df.info()` that it is now datetime64.

2 values ("not sure") couldn't be parsed and became NaT. I kept these rows and
left the date as missing instead of inventing one, because the rest of each
row is valid and the date isn't critical for outreach. Row count is still 71.

### Day 6: **Data Quality Summary**

**Starting dataset:** 86 rows, 7 columns

**Issues found:**
- Missing values: 3 in Applicant Name, 6 in Email, 6 in University, 4 in Status
- 3 exact duplicate rows, plus 3 more duplicate applications found by
  matching Email (same email in different casing)
- Domain Applied had 18 spelling/casing variations for only 5 real domains
- Status had 8 variations for 3 real values
- Applicant Name and Email had inconsistent casing and extra spaces
- Application Date was stored as text in three formats, including 2 unreadable
  values ("not sure")
- Phone numbers contained inconsistent dash/space formatting

**Actions taken:**
- Dropped 3 rows with missing Applicant Name (unusable without identification)
- Dropped 6 rows with missing Email (the contact method and the key for
  detecting duplicates)
- Filled missing University with "Not Specified" (non-critical field)
- Filled missing Status with "Under Review" (default for unprocessed applications)
- Removed 6 duplicate rows total (3 exact, 3 by matching Email)
- Standardized Domain Applied into 5 categories using a mapping dictionary
- Standardized Status into 3 values, and Names to Title Case
- Converted Application Date to datetime64; the 2 unreadable values became
  NaT and were kept, since the rest of those rows is valid
- Removed dashes and spaces from Phone (kept as text to preserve leading zeros)

**Final dataset:** 71 rows, 7 columns (15 rows removed in total), no missing
values in critical fields (name, email), no duplicates, consistent formatting.
Only Application Date has 2 missing values.
